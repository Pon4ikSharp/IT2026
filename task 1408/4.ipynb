1. Зчитати файл бойового донесення та вивести кількість втрат противника.
2. Створити файл `summary.txt`, у якому підсумувати ситуацію за день.
3. Зчитати CSV-файл зі втратами по напрямках і вивести дані лише за напрямком "Південь".
4. Додати лог-файл, куди записується кожен факт зчитування.


```python
import pandas as pd
data = pd.read_csv("boiove_donesennia.csv")
vtraty = data["Втрати особового складу"].sum()
print("Кількість втрат противника:", vtraty)
```

```python
vtraty_os = data["Втрати особового складу"].sum()
vtraty_teh = data["Втрати техніки"].sum()

with open("summary.txt", "w", encoding="utf-8") as f:
    f.write(f"Кількість донесень: {len(data)}\n")
    f.write(f"Втрати о/с: {vtraty_os}\n")
    f.write(f"Втрати техніки: {vtraty_teh}\n")
```
